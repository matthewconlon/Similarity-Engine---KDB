# 03 - Peak KDB Similarity

This notebook builds the **Peak Kevin De Bruyne** archetype and finds the most similar players.


## Archetype logic

Two supported approaches:
- best single season
- average of best three seasons


```python
from pathlib import Path
import sys

import pandas as pd

sys.path.append(str(Path('..').resolve() / 'src'))
from create_peak_kdb_profile import create_peak_profile
from similarity_engine import score_similarity
from explainability import create_explanations
from pca_visualisation import build_pca_plot

features_file = Path('../data/player_season_features.csv')
profile_file = Path('../data/peak_kdb_profile.csv')
output_dir = Path('../output')

profile = create_peak_profile(features_file, profile_file, mode='best_three_average')
features_df = pd.read_csv(features_file)
rankings = score_similarity(features_df, profile, metric='cosine')
rankings.head(25)
```

```python
explanations = create_explanations(features_df, profile, rankings, top_n=25)
explanations.head()
```
